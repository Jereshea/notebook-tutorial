# Basic statistics
Computes summary statistics with the standard library only.

**Expected result:** passes as-is.

```python
import statistics

readings = [12.1, 11.8, 12.4, 12.0, 11.9, 12.3]
```

```python
print('mean  :', round(statistics.mean(readings), 3))
print('median:', statistics.median(readings))
print('stdev :', round(statistics.stdev(readings), 3))
```
